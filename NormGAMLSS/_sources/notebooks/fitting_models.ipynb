# Fitting GAMLSS Models

### gamlss_main.<br>**Gamlss**

#### gamlss_main.Gamlss.<br>**__init__**

```python
Gamlss(model_name, x_vals, y_val)
```

&nbsp;&nbsp;&nbsp;&nbsp;Initialize a GAMLSS model instance.

&nbsp;&nbsp;&nbsp;&nbsp;<span style="background-color: #f5f5f5; padding: 4px 8px; display: inline-block; width: 90%;">**Parameters:**</span>

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**model_name** : ***str*** <br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Name for the R model object

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**x_vals** : ***list of str*** <br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;List of predictor variable names

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**y_val** : ***str*** <br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Name of the target variable

<br>

**Examples:**


````python
from normgamlss.gamlss_main import Gamlss

# Defining x and y variables for GAMLSS
y_val = 'y_thickness1'
x_vals = ['age', 'sex', 'site', 'feature_volume1']
columns = x_vals.copy()
columns.extend([y_val])
# Selecting the columns for the training data
df_train = train_data1.loc[:, columns]

# Create an instance of the Gamlss class
model_name = 'model_shash'
gamlss_model = Gamlss(model_name, x_vals, y_val)
````

#### gamlss_main.Gamlss.<br>**fit**

```python
fit(r_code, data)
```

&nbsp;&nbsp;&nbsp;&nbsp;Fits a GAMLSS model using the provided R code. The R code should contain the complete gamlss() function call with formulas, family, and method specifications.

&nbsp;&nbsp;&nbsp;&nbsp;<span style="background-color: #f5f5f5; padding: 4px 8px; display: inline-block; width: 90%;">**Parameters:**</span>

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**r_code** : ***str*** <br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;R code as a string, defining the model formula and other arguments

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**data** : ***pandas.DataFrame*** <br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;DataFrame containing the data for fitting the model

&nbsp;&nbsp;&nbsp;&nbsp;<span style="background-color: #f5f5f5; padding: 4px 8px; display: inline-block; width: 90%;">**Returns:**</span>

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;**None** : ***None*** <br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Fits the model and stores it in R environment

<br>

**Examples:**

````python
# Fitting GAMLSS model
gamlss_model.fit(
    r_code="""
    model <- gamlss(y ~ pvc(age, by = as.factor(sex)) + re(random=~1|as.factor(site)),
        sigma.formula = ~ pvc(age) + re(random=~1|as.factor(site)),
        nu.formula = ~ 1,
        tau.formula = ~ 1,
        family = SHASH(),
        method = mixed(10, 100)
    )
    """,
    data=df_train
)
# family = SHASH(), GG()
# method = RS(2), CG(10), mixed(2,10)
# mixed(10, 100): 10 RS iterations, then up to 100 CG iterations. With SHASH, RS alone
# can need hundreds of iterations here (nu and tau move slowly); check gamlss_model's
# summary or the 'Converged' columns of the pipelines.
# pvc(x, by=<factor>) fits a separate smooth term per factor level (here: per sex).
````
